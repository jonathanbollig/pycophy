### List Comprehensions
List comprehensions are a concise and elegant way to create lists. The basic syntax is `[expression for item in iterable]`. They are often more efficient and easier to read than traditional loops for creating lists.

For example, to create a list of squares from 0 to 9:
`squares = [x*x for x in range(10)]`
`print(squares) # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]`

**Task:** You are given a list `temperatures_celsius = [0, 10, 20, 30, 40]`. Use a list comprehension to create a new list, `temperatures_fahrenheit`, that contains each temperature converted to Fahrenheit. The conversion formula is: $F = (C \times \frac{9}{5}) + 32$.

```python
temperatures_celsius = [0, 10, 20, 30, 40]

# Solution
temperatures_fahrenheit = [(c * 9/5) + 32 for c in temperatures_celsius]
print(temperatures_fahrenheit)
```
