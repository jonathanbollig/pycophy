```python

# Filtering strings
mixed_data = [1, 4, 3.0, "Hello", 2, "three", None, True]

print("Strings:")
for cake in mixed_data:
    if type(cake) == str:
        print(cake)

print()
print("Other type of data:")
for cake in mixed_data:
    if type(cake) != str:
        print(cake)
```
