```python

# two types
import numpy as np
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# only y data

# y and x data
plt.plot(x, y)
plt.show()
```

```python

# More data
y2 = np.cos(x)

# plot both
plt.plot(x, y, label="y", c="black", linestyle=":")
plt.plot(x, y2, label="y2")

# add axis labels
plt.xlabel("x")
plt.ylabel("Cool")

# add grid
plt.grid()

# add title
plt.title("Title")
# add label
plt.legend()

# change c

# change linestyle

# mention scatter
plt.show()
```
