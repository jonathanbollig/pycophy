```python
# prelimenary:
from time import time
import numpy as np

np_data = np.arange(1e8)
list_data = list(np_data) # ignore numpy here
# list_data = [1, 2, 3, ...., 1e8]
```

```python
# list_data --> list with [1, 2, 3, ...., 1e8]

# list
start = time()
list_new_data = []
for i in list_data:
    list_new_data.append(i**2)
list_squared_sum = sum(list_new_data)
print("Time for list: " + str(time()-start))
# print(list_squared_sum)
```

```python
# np_data --> numpy format 

# numpy
start = time()
np_squared_sum = np.sum(np_data**2)
print("Time for numpy: " + str(time()-start))
# print(np_squared_sum)
```

```python

# import 
import numpy as np
my_list = [1, 2, 3, 4, 5]
# array()
my_array = np.array(my_list)
# tolist()
new_list = my_array.tolist() 
print(new_list)
```

```python

# mixed list
mixed_list = [3, 2.4, "hi"]
mixed_array = np.array(mixed_list)
print(mixed_array)
```

```python

# Similarities
# print both
print(my_array)
print(my_list)
# index both
print(my_array[0])
print(my_list[0])

# slice both
print(my_array[1:3])
print(my_list[1:3])

# single set value both
my_array[0]=5
my_list[0]=5
print(my_array)
print(my_list)
```

```python

# loop through both
for i in my_array:
    print(i)
```

```python

# Differences
# addition 
print(my_list+my_list)
print(my_array+my_array)
# multiplication
print(3*my_array)

# more examples:
print(my_array**2)
print(my_array % 2)
print(my_array / 2)
print(np.sin(np.tan(np.cos(my_array))))
```
