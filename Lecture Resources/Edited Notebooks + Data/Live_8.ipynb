```python
# prelimenary:
from time import time
import numpy as np

np_data = np.arange(1e8)
list_data = list(np_data) # ignore numpy here
# list_data = [1, 2, 3, ...., 1e8]
```

```python
# numpy
start = time()
np_squared_sum = np.sum(np_data**2)
print("Time for numpy: " + str(time()-start))
print(np_squared_sum)
```

```python
# list
start = time()
list_new_data = []
for i in list_data:
    list_new_data.append(i**2)
list_squared_sum = sum(list_new_data)
print("Time for list: " + str(time()-start))
print(list_squared_sum)
```
