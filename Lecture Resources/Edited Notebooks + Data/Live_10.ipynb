```python
import numpy as np

my_list = [[1, 2, 3], 
           [4, 5, 6]]
my_array = np.array(my_list)

my_array = np.array([[1, 2, 3], 
                     [4, 5, 6]])

another_array = np.zeros((2, 2, 2, 2))
print(another_array)
```

```python

# indexing
# without coma --> first dimension
my_array = np.array([[1, 2, 3], 
                     [4, 5, 6]])
print(my_array[0, 1])
# coma access next dimension
```

```python

# slicing
# first dimension
print(my_array[0, :])
# second dimension
print(my_array[:, 0])

# : without more
```

```python

my_array = np.array([[1, 2, 3], 
                     [4, 5, 6]])
# Useful stuff
print(my_array.T)
# transpose

# flatten
print(my_array.flatten())

# concatenate(())
print(np.concatenate((my_array, my_array)))
```
