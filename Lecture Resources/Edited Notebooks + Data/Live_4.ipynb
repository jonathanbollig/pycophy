```python

my_list = [100, 50, 0, 99, 4, 42]
# print, print+index
# index from back
# slicing
# from beginning / end
print(my_list[1:-1])
# complete list
print(my_list)
```

```python

my_list = [100, 50, 0, 99, 4, 42]
# change element in list
my_list[0] = 1
print(my_list)
# change multiple elements in list
my_list[-2:] = [40, 420, 100]
print(my_list)
```

```python

list1 = [1, 2, 3, 4]
list2 = ["a", "b", "c"]

# arithmetic operators on lists

# +
print(list1 + list2)
# *
print(3*list2)
# len
print(len(list2))
```

```python

my_list = [100, 50, 0, 99, 4, 42]
# append
my_list.append(99)
print(my_list)
# min 
print(min(my_list))
# max
# sum
print(sum(my_list))
```
