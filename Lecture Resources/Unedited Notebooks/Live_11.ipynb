```python

# better way: pandas + numpy
import numpy as np
import pandas as pd

filename = "data\\daily_temperature_cleaned.txt"
data = pd.read_table(filename, delimiter=',').to_numpy()
print(data)
```

```python

# try and error:
filename = "data\\daily_temperature.txt"
data = pd.read_table(filename).to_numpy()
print(data)
```

```python

# add skiprow
filename = "data\\daily_temperature.txt"
data = pd.read_table(filename, skiprows=22).to_numpy()
print(data)
```

```python

# add delimiter
filename = "data\\daily_temperature.txt"
data = pd.read_table(filename, skiprows=22, delimiter='\\s+').to_numpy()
# print(data)
# print(data[0])
```

```python

# select relevant data
filename = "data\\daily_temperature.txt"
data = pd.read_table(filename, skiprows=22, delimiter='\\s+').to_numpy()[:, [0, 5]]
print(data)
```

```python

# save data human readable
save_filename = "data\\daily_temperature_cleaned2.txt"
np.savetxt(save_filename, data, fmt='%.3f', delimiter=",")
```
