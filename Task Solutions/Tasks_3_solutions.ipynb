It should state what beverages you are allowed to buy (in Germany wine and beer from 16 years of age, and everything else from 18 years of age).

Tip: Think about type conversion. 

```python
age = input("Please insert your age")

# Solution
age = int(age)
if age < 16:
    print("You cannot buy alcohol")
elif age < 18:
    print("You can buy beer and wine")
else:
    print("You can buy all alcohol")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 3.3**</span> Write some code, that lets the user input a day of the week. If the day is a day of the weekend, print a happy message. If not, print a sad message. Does capitalizing the input make any difference?

Extra: If the user doesn't enter a week day, the code should tell, that the input was not a day. 

```python
# solution
day = input("Give me a weekday")

if day == "Saturday" or day == "Sunday":
    print("Yaaay, it's weekend")
elif day == "Monday" or day == "Tuesday" or day == "Wednesday" or day == "Thursday" or day == "Friday":
    print("Oh no I have to work :(")
else:
    print("Sorry, you didn't enter a day of the week.")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 3.5**</span> Complete the program that asks the user to input two numbers and an operator (+, -, *, /). Use if-elif-else statements to calculate and print the result. If the operator is invalid, print an error message. Make sure, your code does not divide by zero. 

<span style="font-size:12px; color:gray;">
Credits to perplexity for this task.
</span>

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operator = input("Enter an operator (+, -, *, /): ")

# solution
if operator == "+":
    print(f"Result: {num1 + num2}")
elif operator == "-":
    print(f"Result: {num1 - num2}")
elif operator == "*":
    print(f"Result: {num1 * num2}")
elif operator == "/":
    if num2 != 0:
        print(f"Result: {num1 / num2}")
    else:
        print("Error: Division by zero")
else:
    print("Invalid operator")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 3.6**</span> Repeat task 3.3 with the weekdays. But before, make sure, as much as possible, that user input is "sanitized". This means `"Monday"`, `"MONDAY"`, `" Monday     "` should all be handled properly in the if statement.

```python
# solution
day = input("Give me a weekday")

# remove whitespaces
day = day.strip()

# Make first letter Capital, the other small
day = day.capitalize()

if day == "Saturday" or day == "Sunday":
    print("Yaaay, it's weekend")
elif day == "Monday" or day == "Tuesday" or day == "Wednesday" or day == "Thursday" or day == "Friday":
    print("Oh no I have to work :(")
else:
    print("Sorry, you didn't enter a day of the week.")
```
