<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 7.1**</span> From `time` import `time` and `sleep`. The `time()` function returns the current time in seconds ([unix epoch time](https://www.unixtimestamp.com/)). The `sleep()` function makes the code sleep for `n` seconds. Test, how accurate the `sleep` and `time` methods are.

```python
# solution
from time import time, sleep

start_time = time()
sleep(1)
end_time = time()
print(f"Time elapsed: {end_time-start_time}")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 7.2**</span> Import the `math` module and use it to calculate the factorial of 7. The function for this is `math.factorial()`.

Extra: write your own function, that calculates the factorial.

```python
# Solution
import math

result = math.factorial(7)
print(f"The factorial of 7 is: {result}")

# Extra:
def factorial(number):
    result = 1
    for i in range(1, number+1):  # starting from 1 (not 0) and going until number (including)
        result = result*i
    return result
# Note: This can also be done elegantly with a recursive function (more on this later)

result = factorial(7)
print(f"The factorial of 7 is: {result}")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 7.3**</span> From the `math` module, import the `radians` and `cos` functions directly. Then, calculate the cosine of 90 degrees. (Note: `cos` expects the angle in radians, so you'll need to convert it first!). Is the number, what you would expect?

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
# Solution
from math import radians, cos

angle_degrees = 90
angle_rad = radians(angle_degrees)
result = cos(angle_rad)

print(f"The cosine of 90 degrees is: {result}")
# Note: The result is not 0, but a very small number close to zero, due to floating-point precision.
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 7.5**</span> Implement task 7.3 recursivly. 

```python
# Solution
def recursive_factorial(number):
    if number == 1:
        return 1
    return number * recursive_factorial(number - 1)

print(recursive_factorial(7))
```
