<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 4.2**</span> Create a list containing the following numbers: `23, 45, 12, 58, 6`. Write code to calculate and print the average of these numbers. 

Tip: Use `sum()` and `len()`!

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
# Solution
numbers = [23, 45, 12, 58, 6]
average = sum(numbers) / len(numbers)
print("The average is:", average)
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 4.3**</span> Use the list named `planets` with the names of the eight planets in our solar system (in order from the sun). Do the following:
- Print the third planet from the sun.
- Print the last two planets in the list using slicing.
- Add the dwarf planet "Pluto" to the end of your list and print the new, updated list.

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]

# Solution
print("The third planet is:", planets[2])
print("The last two planets are:", planets[-2:])

planets.append("Pluto")
print("The list with Pluto:", planets)
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 4.4**</span> You have a list of letters `letters = ['a', 'b', 'c', 'f', 'g']`. You noticed that `'d'` and `'e'` are missing. Use slicing to replace the end of the list to correct it. The final list should be `['a', 'b', 'c', 'd', 'e', 'f', 'g']`.

```python
letters = ['a', 'b', 'c', 'f', 'g']

# Solution
letters[3:] = ['d', 'e', 'f', 'g']
print(letters)
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 4.6**</span> Write a piece of code, that checks if a matrix has equal side lengths (assume all lists in the matrix have the same length).

Apply the code on matrix a and b (copy paste the code).

```python
matrix_a = [["a", "b", "c"], 
            [1, 9, True]]

matrix_b = [[20, 10.1, 2], 
            [3, 9, 0.1], 
            [1000000, 1, 0]]

# solution
if len(matrix_a) == len(matrix_a[0]):
    print("This is a equally dimensioned matrix")
else:
    print("The sides of the matrix have different lengths")

if len(matrix_b) == len(matrix_b[0]):
    print("This is a equally dimensioned matrix")
else:
    print("The sides of the matrix have different lengths")
    
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 4.7**</span> Write code, that changes the first an second line of this matrix. 

Tip: Use a variable, to intermediatly save a line.

```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
print(matrix)

# solution
line_1 = matrix[0]
matrix[0] = matrix[1]
matrix[1] = line_1
print(matrix)
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 4.8**</span> Advanced: Instead of changing rows, change the first and second column. 

```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
print(matrix)

# solution
column_1 = [matrix[0][0], matrix[1][0], matrix[2][0]]
matrix[0][0] = matrix[0][1]
matrix[1][0] = matrix[1][1]
matrix[2][0] = matrix[2][1]
matrix[0][1] = column_1[0]
matrix[1][1] = column_1[1]
matrix[2][1] = column_1[2]
print(matrix)
```
