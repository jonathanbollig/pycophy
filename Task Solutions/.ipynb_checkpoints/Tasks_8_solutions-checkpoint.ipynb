<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 8.2**</span> A car accelerates at 1 m/s from a starting speed of 5 m/s. The formula for the speed as a function of time is: $v(t)=v_0+at$

Given is an array of times in seconds. Calculate the speed of the car at these times. 

```python
times = np.linspace(0, 5, 10)  # s
v_0 = 5  # m/s
a = 1  # m/s/s
print(np.round(times, 2))

# solution
speeds = v_0 + a * times
print(np.round(speeds, 2))
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 8.3**</span> You have a list of temperatures in Celsius. Convert this list to a NumPy array and then use a single line of code to convert all temperatures to Fahrenheit. The formula is: $F = C \times \frac{9}{5} + 32$.

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
celsius_temps = [0, 10, 20, 30, 100]

# Solution
celsius_array = np.array(celsius_temps)
fahrenheit_array = celsius_array * (9/5) + 32

print(f"Celsius: {celsius_array}")
print(f"Fahrenheit: {fahrenheit_array}")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 8.4**</span> Given two NumPy arrays representing the masses and velocities of five objects, calculate the kinetic energy ($E_k = \frac{1}{2}mv^2$) for each object. Store the results in a new array called `kinetic_energies`.

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
masses = np.array([10, 20, 5, 2, 50])      # in kg
velocities = np.array([2, 3, 10, 20, 4]) # in m/s

# Solution
kinetic_energies = 0.5 * masses * velocities**2
print(f"Kinetic Energies (Joules): {kinetic_energies}")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 8.5**</span> NumPy has its own versions of mathematical functions that operate element-wise on arrays. For example, `np.sum()`, `np.mean()`, `np.max()`, `np.std()`, and `np.min()`. Use the array from the previous task to calculate the total, average and standard deviation of the  kinetic energy.

```python
# Solution
total_energy = np.sum(kinetic_energies)
average_energy = np.mean(kinetic_energies)
energy_std = np.std(kinetic_energies)

print(f"Total Kinetic Energy: {total_energy} J")
print(f"Average Kinetic Energy: {average_energy} +- {round(energy_std)} J")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 8.6**</span> In physics, you often want to evaluate a function over a range of values. Create a NumPy array named `x` that contains 100 points evenly spaced from 0 to $2\pi$. Then, calculate the sine of each of these points and store it in an array named `y`. (Use `np.pi` for $\pi$ and `np.sin()` for the sine function). Print the first five values of each array.

```python
# Solution
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Let's print the first 5 values of each to check
print("First 5 x values:", x[:5])
print("First 5 y values:", y[:5])
```
