<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 10.6**</span> An experiment records the temperature at 3 different sensor locations over 10 seconds. The data is stored in a 10x3 array. The first column is Sensor 1, second is Sensor 2, etc. 

Print the data, to get an idea how it looks like. Plot the temperature readings of Sensor 1 (y-axis) vs. Sensor 3 (x-axis) to see if their readings are correlated. Add labels and a title.

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
# Generating some sample data
experiment_data = np.array(np.linspace(0, 5, 10) + [np.random.rand(10) * 0.5, np.random.rand(10) * 0.5, np.random.rand(10) * 0.5]).T

# Solution
print(experiment_data)

s1 = experiment_data[:, 0]
s3 = experiment_data[:, 2]

plt.plot(s3, s1, 'o-') # 'o-' plots points and a line
plt.title("Sensor 1 vs. Sensor 3 Correlation")
plt.xlabel("Sensor 3 Temperature (°C)")
plt.ylabel("Sensor 1 Temperature (°C)")
plt.grid(True)
plt.show()
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 10.7**</span> A common task in physics is to calculate the distance of particles from the origin. Given the `particle_coords` array from a previous task, calculate the distance $d = \sqrt{x^2 + y^2}$ for *each* particle. 

*Hint:* First, extract the x and y coordinates into separate 1D arrays. Then, use element-wise operations.

```python
particle_coords = np.array([[1.2, 0.5],
                            [0.8, -0.3],
                            [-0.1, 1.5]])
# Solution
x_coords = particle_coords[:, 0]
y_coords = particle_coords[:, 1]

distances = np.sqrt(x_coords**2 + y_coords**2)
print(f"Distances from origin: {np.round(distances, 2)}")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 10.8**</span> Model a simple 2D electric potential, given by the formula $V(x, y) = \sqrt{x^2 + y^2}$, which is the potential of a point charge at the origin. 

The following was already done:
1. Create 1D `x` and `y` arrays from -5 to 5 with 100 points each.
2. Use `np.meshgrid(x, y)` to create 2D coordinate grids `X` and `Y`.

Now do the following:

3. Use `imshow` to get an idea, what `X` and `Y` look like. Then comment out the line, so the plots aren't shown permanently.
4. Calculate the potential `V` for every point on the grid using the formula.
5. Use `plt.contourf(X, Y, V)` to visualize the potential field. Use `plt.colorbar()` to add a scale.


<span style="font-size:12px; color:gray;">
Credits to Gemini for this cell (Mainly, I did some work too ;).
</span>

```python
# 1. Create 1D coordinate arrays
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# 2. Create 2D coordinate grids
X, Y = np.meshgrid(x, y)

# Solution
# 3. Imshow the data and comment it out:
# plt.imshow(X)
# plt.show()
# plt.imshow(Y)
# plt.show()

# 4. Calculate the potential V on the grid
V = np.sqrt(X**2 + Y**2)

# 5. Create a filled contour plot
plt.contourf(X, Y, V, levels=15) # levels adds more contour lines
plt.title("Electric Potential of a Point Charge")
plt.xlabel("x-coordinate")
plt.ylabel("y-coordinate")
plt.colorbar(label="Potential V")
# plt.gca().set_aspect('equal', adjustable='box') # Makes the plot square
plt.show()
```
