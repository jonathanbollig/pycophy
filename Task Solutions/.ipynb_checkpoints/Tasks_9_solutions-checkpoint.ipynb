<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 9.2**</span> Given are a set of x- and y-data-arrays. Plot them, and try to guess, what mathematical function describes these points best. Then try to replicate that function in a new array (like the sine in the cheat-sheet). Plot both together, and check if they look similar. 

```python
x = [0, 0.5, 1, 2, 4, 5, 9, 14]
y = [0, 0.7, 0.9, 1.5, 2.2, 2.4, 3.1, 3.8]

# solution
plt.plot(x, y)
plt.show()

y_theory = np.sqrt(x)
plt.plot(x, y)
plt.plot(x, y_theory)
plt.show()
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 9.3**</span> Let's revisit **Task 8.2**. A car accelerates at 1 m/s² from a starting speed of 5 m/s. Its speed is given by $v(t) = v_0 + at$. 

Given is the solution for the times and speeds array. Now create a simple plot of speed (y-axis) vs. time (x-axis).

```python
v_0 = 5  # m/s
a = 1  # m/s^2

times = np.linspace(0, 10, 100)  # s
speeds = v_0 + a * times

# Solution
plt.plot(times, speeds)
plt.show()
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 9.4**</span> Now, improve the plot from **Task 9.2**. Add a title ("Speed of an Accelerating Car"), an x-label ("Time (s)"), and a y-label ("Speed (m/s)"). Also, add a grid to the background to make it easier to read.

```python
# Solution
plt.plot(times, speeds)

plt.title("Speed of an Accelerating Car")
plt.xlabel("Time (s)")
plt.ylabel("Speed (m/s)")
plt.grid(True)

plt.show()
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 9.5**</span> Another car starts at 20 m/s and *decelerates* at -2 m/s².

Plot the speed of **both** cars on the same graph. 
- The first car's line should be blue and solid.
- The second car's line should be red and dashed (`linestyle="--"`).
- Add labels to each plot call (`label="Car 1"`, `label="Car 2"`).
- Finally, display a legend using `plt.legend()`.

```python
speeds2 = 20 - 2 * times

# Solution

plt.plot(times, speeds, label="Car 1", color="blue", linestyle="-")
plt.plot(times, speeds2, label="Car 2", color="red", linestyle="--")

plt.title("Comparing Car Speeds")
plt.xlabel("Time (s)")
plt.ylabel("Speed (m/s)")
plt.grid(True)
plt.legend()

plt.show()
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 9.6**</span> In **Task 8.5**, we calculated the kinetic energy for five objects. Instead of a line plot, a scatter plot is often better for individual data points. Use `plt.scatter()` to plot the `kinetic_energies` (y-axis) against the `velocities` (x-axis). Add appropriate labels and a title.

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
masses = np.array([10, 20, 5, 2, 50])      # in kg
velocities = np.array([2, 3, 10, 20, 4]) # in m/s
kinetic_energies = 0.5 * masses * velocities**2

# Solution
plt.scatter(velocities, kinetic_energies)

plt.title("Kinetic Energy vs. Velocity")
plt.xlabel("Velocity (m/s)")
plt.ylabel("Kinetic Energy (J)")
plt.grid(True)

plt.show()
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 9.7**</span> Recreate the Sine and Cosine plot from the cheat-sheet, but this time, use the object-oriented approach (`fig, ax`).

```python
x_data = np.linspace(0, 2 * np.pi, 100)
y1_data = np.sin(x_data)
y2_data = np.cos(x_data)

# Solution
# 1. Create the figure and axes objects
fig, ax = plt.subplots()

# 2. Use the 'ax' object for plotting
ax.plot(x_data, y1_data, label="Sine Wave", color="blue", linestyle="-")
ax.plot(x_data, y2_data, label="Cosine Wave", color="red", linestyle="--")

# 3. Use the 'set_' methods to add decorations
ax.set_title("Sine and Cosine Functions (OO-Style)")
ax.set_xlabel("Angle [rad]")
ax.set_ylabel("Value")
ax.grid(True)
ax.legend()

# 4. Show the plot as usual
plt.show()
```
