<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 11.3**</span> To avoid repeating code, write a Python **function** `load_temp_data(filename)` that takes a filename (e.g., `daily_temperature.txt`) as input. The function should:
1. Construct the full relative path (e.g., `data/filename`).
2. Load the data using pandas, skipping the 22 header rows.
3. Convert it to a NumPy array.
4. **Return** the NumPy array.

Test your function by loading `daily_temperature_high.txt` and `daily_temperature_low.txt`.

```python
# Solution
def load_temp_data(filename):
    path = f"data/{filename}"
    data = pd.read_csv(path, sep='\\s+', skiprows=22).to_numpy()[:,[0, 5]]
    return data

# Test the function
high_temp_data = load_temp_data("daily_temperature_high.txt")
low_temp_data = load_temp_data("daily_temperature_low.txt")
print(f"Loaded high temp data with shape: {high_temp_data.shape}")
print(f"Loaded low temp data with shape: {low_temp_data.shape}")
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 11.4**</span> The data is currently a 2D array containing pairs of `year` and `anomaly`. Because the use is easier, extract the values in 1D arrays **year** (column 0) and the **temperature anomaly** (column 1).

Using the `avg_temp_data` array you loaded earlier, use NumPy **slicing** to extract these two columns. Store them in new 1D arrays called `years` and `anomalies`.

```python
# Solution
years = avg_temp_data[:, 0]       # All rows, column index 1
anomalies = avg_temp_data[:, 1]  # All rows, the last column

print("First 5 years:", years[:5])
print("First 5 anomalies:", anomalies[:5])
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 11.5**</span> Now, let's visualize the results! <span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 11.6**</span> You have now extracted the most important information. Create a new 2D NumPy array called `cleaned_data` that contains just two columns: the `years` and the `anomalies`.

Then, **save** this `cleaned_data` to a new file named `cleaned_temperature_data.csv` inside the `data` folder. Use a comma as the delimiter and format the numbers to have 4 decimal places.

```python
# Solution
# np.stack is used to join the 1D arrays back into a 2D array
cleaned_data = np.stack((years, anomalies), axis=1)

save_path = "data/cleaned_temperature_data.csv"
header_text = "Year,TemperatureAnomaly_C"

np.savetxt(save_path, cleaned_data, delimiter=",", fmt='%.4f', header=header_text)

print(f"Saved cleaned data with shape {cleaned_data.shape} to {save_path}")
```
