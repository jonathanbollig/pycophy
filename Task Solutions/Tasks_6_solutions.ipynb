<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 6.2**</span> Some Physics: We have four point masses. All have the same weight. Calculate the center of mass.

Given are their positions in x and y, in two respective lists.

```python
x_positions = [2, 10, 0, 2]
y_positions = [4, 4, -0.5, 8]

# solution
def center_of_mass(x_pos, y_pos):
    x_center = sum(x_pos) / len(x_pos)
    y_center = sum(y_pos) / len(y_pos)
    return x_center, y_center

print(center_of_mass(x_positions, y_positions))
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 6.5**</span> In the code block from your presentation (`Tasks_4_lists_tuples`), you had to copy and paste code to check if two different matrices had equal side lengths. Rewrite that logic into a single, reusable function named `check_matrix_dimensions`. The function should take a matrix (list of lists) as input and print whether it is equally dimensioned or not. Then, call this function for both `matrix_a` and `matrix_b`.

```python
matrix_a = [["a", "b", "c"], 
            [1, 9, True]]

matrix_b = [[20, 10.1, 2], 
            [3, 9, 0.1], 
            [100000, 1, 0]]

# Solution
def check_matrix_dimensions(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    if rows == cols:
        print("This is an equally dimensioned matrix (square).")
    else:
        print("The sides of the matrix have different lengths.")

print("Checking Matrix A:")
check_matrix_dimensions(matrix_a)

print("\nChecking Matrix B:")
check_matrix_dimensions(matrix_b)
```

<span style="background-color: #f8bbd0; color: black; padding: 2px 5px; border-radius: 3px;">**Task 6.6**</span> Write a function `find_max` that takes a list of numbers and returns the largest number in the list without using the built-in `max()` function. You can use a loop to solve this.

<span style="font-size:12px; color:gray;">
Credits to Gemini for this task.
</span>

```python
data = [1, 5, 2, 9, 4]

# Solution
def find_max(numbers):
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest

print(f"The largest number is: {find_max(data)}")
```
