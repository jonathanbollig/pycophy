**SciPy** (Scientific Python) is a library that builds on NumPy and provides a large number of functions for scientific and technical computing. We will focus on its tools for curve fitting and its collection of physical constants.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import scipy.constants as const
import os
```

**Plotting Experimental Data**: When plotting measured data points, it's best to show them as individual markers without a connecting line. This is done by setting `linestyle=''` (or `'none'`) and specifying a `marker`.

```python
x_data = np.array([0, 1, 2, 3, 4])
y_data = np.array([0.1, 1.2, 1.9, 3.3, 4.1])

plt.plot(x_data, y_data, marker='o', linestyle='', label='Measured Data')
plt.legend()
plt.show()
```

**Polynomial Fitting with `numpy.polyfit`**: For data that follows a polynomial trend ($y = ax^n + bx^{n-1} + ... + c$), `np.polyfit` is the easiest tool. It returns the coefficients of the polynomial.

```python
x_poly = np.linspace(0, 4, 20)
y_poly = x_poly**2 - 2*x_poly + 1 + (np.random.rand(20) - 0.5)

# Fit a 2nd order polynomial (a quadratic)
coeffs = np.polyfit(x_poly, y_poly, 2)
print(f"Coefficients (a, b, c): {np.round(coeffs)}")
```

**General Fitting with `scipy.optimize.curve_fit`**: For any other function (e.g., exponential, sinusoidal), use `curve_fit`. You must first define a Python function for your model. It returns the optimal parameters (`popt`) and the covariance matrix (`pcov`).

```python
# 1. Define the model function
def exponential_decay(x, A, k):
    return A * np.exp(-k * x)

x_exp = np.linspace(0, 4, 20)
y_exp = 5 * np.exp(-0.5 * x_exp) + (np.random.rand(20) - 0.5)

# 2. Call curve_fit
popt, pcov = curve_fit(exponential_decay, x_exp, y_exp)
print(f"Fitted parameters (A, k): {popt}")
```

### Finding the Uncertainty of Fitted Parameters

In experimental physics, a result is meaningless without its uncertainty. The `curve_fit` function provides the **covariance matrix (`pcov`)** for exactly this purpose. 

The diagonal elements of this matrix are the *variances* of the fitted parameters. To get the **standard error** (the uncertainty), you just need to take the square root of these diagonal elements.
