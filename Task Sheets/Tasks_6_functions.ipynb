A **function** is a reusable block of code that performs a specific action. You define a function using the `def` keyword.

One or more arguments come in `()`, separated by commas. 

```python
def greet2(name1, name2):
    print(f"Hello {name1} and {name2}!")

# Calls the function
greet2("Romeo", "Julia")
```

The **`return` statement** is used to send one or multiple values back from the function. This allows you to store the function's result in a variable. Statements after `return` are not executed

```python
def square(number):
    return number ** 2
    print("Hahaha this is a useless print statement")  # this will never be executed.

result = square(5)
print(f"The result is: {result}")
```

Functions can also return **multiple values**.

```python
from math import sqrt, atan2

def cartesian_to_polar(x, y):
    r = sqrt(x**2 + y**2)
    phi = atan2(y, x) # Note: atan2 is better than atan for this
    return r, phi

radius, angle = cartesian_to_polar(10, 5)
print(f"Radius: {radius}")
print(f"Angle (radians): {angle}")
```

### Optional Arguments

Let's take a look at the `round()` function, and how we can use it:

```python
floating = 1.23456
print(round(floating))
print(round(floating, 2))
```

We can use `round` with one argument: The number we want to round, or with multiple arguments, the number we want to round, and the number of decimal places.

This means, the second argument (number of decimal places) is optional. We can give it, but we don't have to.

Let's see how we can define a function with an optional argument:

```python
def print_info(name, age=None):
    if age is not None:  # to check if a value is (not) None
        print(f"Your name is {name} and you are {age} years old.")
    else:
        print(f"Your name is {name} and I don't know how old you are.")

print_info("Dieter", 36)
print()
print_info("Dieter")
```

We see: To create a function with an optional argument, we need to set a default value. If the optional value is not passed, the default value is used. In our case this means, `age` is set to the value `None`. Then, the code in the function can work according to the value.
