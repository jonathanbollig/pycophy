```python
import scipy 
import scipy.constants as c
import numpy as np

# %matplotlib widget
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (4,3)
plt.rcParams["axes.grid"] = True
```

```python
x = np.linspace(-10, 10, 100)
y_theory = x**3 + 5 *x**2 - 50*x
y_measurement = y_theory + 100*(np.random.rand(len(x)) - 0.5)

plt.plot(x, y_measurement, marker="x", linestyle="")
plt.show()

np.savetxt("data\\task12_1.txt",np.array([x, y_measurement]).T, fmt="%.3f")
print(y_measurement)
```
