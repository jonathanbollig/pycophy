A **`for` loops** is used to iterate over a sequence like a list, tuple, or other iterable objects. It repeatedly executes the code block for each item in the sequence.

````python
# The loop assigns each item from my_list to the variable 'element' one by one
my_list = [100, 50, 0, 99]
for element in my_list:
    print(element)
````

The **`range()` function** generates a sequence of numbers, which is very useful for looping a specific number of times. `range(n)` generates numbers from 0 up to, but not including, `n`.

````python
# Prints the numbers 0, 1, 2, 3, 4
for i in range(5):
    print(i)
````

The **`zip()` function** is used to iterate over two or more iterables at the same time. It pairs up the elements from each iterable.

````python
names = ["Dieter", "Felix", "Mareike", "Lea"]
ages = [43, 4, 23, 29]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old.")
````

A **`while` loops** executes a block of code as long as a specified condition is true. This is useful when the number of iterations is not known in advance.

````python
# This loop will continue asking for input until the user enters a number greater than 20.
number = input("Enter a positive number > 20: ")
while int(number) <= 20:
    number = input("Try again: ")
print("Success")
````

**`break` and `continue`**
- **`break`**: Exits the loop immediately.
- **`continue`**: Skips the rest of the current iteration and moves to the beginning of the next one.

````python
# Break example: stops when it finds a number > 6
print("Break example:")
data = [1, 4, 2, 9, 4, 5, 8]
for i in data:
    if i > 6:
        break
    print(i)

# Continue example: skips printing even numbers
print("\nContinue example:")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
````

### Nested Loops for matrices
You can use one loop inside another loop. This is called a **nested loop**. It is very useful for working with multi-dimensional data structures, like a list of lists (which can represent a matrix or a grid).

**Example:**
```python
for i in range(2):       # Outer loop runs 2 times
    for j in ['a', 'b']: # Inner loop runs 2 times for each outer loop iteration
        print(i, j)
```
**Output:**
```
0 a
0 b
1 a
1 b
```
