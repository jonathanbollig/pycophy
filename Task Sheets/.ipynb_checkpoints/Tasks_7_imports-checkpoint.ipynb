The **`import` statement** is used to bring modules (files containing Python code) into your current script. This allows you to use functions and variables defined in those modules.

```python
# Import the entire 'math' module
import math

# You must use the module name to access its functions
print(math.sqrt(16))
```

Using **`from ... import ...`** lets you import specific functions or variables directly, so you don't need to use the module name.

```python
# Import only the 'pi' constant and 'sin' function from the math module
from math import pi, sin

# Now you can use them directly
print(f"The value of pi is {pi}")
print(f"The sine of pi/2 is {sin(pi/2)}")
```

You can rename a module or function during import using **`as`**. This is very common for modules with long names or to follow conventions (e.g., `import numpy as np`).

```python
# Import the math module and rename it to 'm'
import math as m
print(m.log(m.e)) # Calculates the natural log of Euler's number, e

# Import a specific function and rename it
from math import log as ln
print(ln(m.e))
```

### Recursive Functions

Functions can call themselves. This means, the function `x` can be called inside the definition of the function `x`. Normally, this would end up in the function calling itself an infinite amount of times. However, we can define a stopping criterion. Below an example. The function prints the last element of a list, and removes it. If the remaining list has a length > 0, the function is called again on the remaining list.

```python
def print_elements(my_list):
    print(my_list[-1])  # print last element
    my_list.pop()  # remove last element
    if len(my_list) > 0:  # stopping criterion
        print_elements(my_list)

print_elements([1, 2, 3, 7, 8, 9])
```
