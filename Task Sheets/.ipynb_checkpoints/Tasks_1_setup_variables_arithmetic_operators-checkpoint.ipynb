Variables are a way to store data:


```python
a = "Hello World"
print(a)
```

"Hello World" is a string. We can also print and save other types of data:
| Name      | What                     | Example       |
|-----------|--------------------------|---------------|
| String    | Text                     | "Hello World" |
| int       | Integer (Number)         | 1, 10, -123   |
| float     | Rational Number          | 3.14, 1.9999  |
| boolean   | True / False             | True, False   |
| Emptyness | Explictly saving nothing | None          |

```python
this_is_a_string = "Hello Mars"
this_is_a_int = 10
this_is_a_float = 9.31
this_is_a_boolean = True
this_is_no_value = None
print(this_is_a_string)
print(this_is_a_int)
print(this_is_a_float)
print(this_is_a_boolean)
print(this_is_no_value)
```
