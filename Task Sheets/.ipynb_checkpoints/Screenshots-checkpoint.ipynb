```python
matrix_a = [["a", "b", "c"], 
            [1,  9,  True ]]

matrix_b = [[20, 10.1, 2], 
            [3,  9,  0.1], 
            [10000, 1, 0]]

# solution
if len(matrix_a) == len(matrix_a[0]):
    print("This is a equally dimensioned matrix")
else:
    print("The sides of the matrix have different lengths")

if len(matrix_b) == len(matrix_b[0]):
    print("This is a equally dimensioned matrix")
else:
    print("The sides of the matrix have different lengths")

# slightly cleaner solution:
for matrix in [matrix_a, matrix_b]:
    if len(matrix) == len(matrix[0]):
        print("This is a equally dimensioned matrix")
    else:
        print("The sides of the matrix have different lengths")
```

```python
def calculate_radius(x, y):
    r = (x**2 + y**2)**(1/2)
    print(r)

calculate_radius(10, 5)
```

```python
from math import sqrt, atan2

def cartesian_to_polar(x, y):
    r = sqrt(x**2 + y**2)
    phi = atan2(x, y)
    return r, phi

r, phi = cartesian_to_polar(10, 5)
print(r)
print(phi)
```

```python
from time import time
import numpy as np

np_data = np.arange(1e8)
list_data = list(np_data) # ignore numpy here

# list_data = [1, 2, 3, ...., 1e8]
# np_data = same thing, but in numpy format
```

```python

# list_data = [1, 2, 3, ...., 1e8]
start = time()

list_new_data = []
for i in list_data:
    list_new_data.append(i**2)
list_squared_sum = sum(list_new_data)

print("Time for list: " + str(time()-start))
```

```python
start = time()

np_squared_sum = np.sum(np_data**2)
    
print("Time for numpy: " + str(time()-start))
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1)
plt.plot(x, y2)
plt.xlabel("x [radians]")
plt.ylabel("y [1]")
plt.grid()
plt.show()
```
