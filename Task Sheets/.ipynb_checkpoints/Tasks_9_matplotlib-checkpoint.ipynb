**Matplotlib** is the most widely used library for creating plots and data visualizations in Python. We almost always import its `pyplot` submodule, which contains the functions for creating plots.

```python
import matplotlib.pyplot as plt
import numpy as np
```

The basic workflow is to call `plt.plot()` for each line you want to add to the figure, and then call `plt.show()` to display the final plot.

```python
x = np.linspace(0, 10, 50)
y = x**2

# You can provide both x and y data
plt.plot(x, y)
plt.show()
```

You can customize your plot by adding labels, a title, a grid, and a legend. For the legend to work, you must add a `label` to each plot.

```python
x_data = np.linspace(0, 2 * np.pi, 100)
y1_data = np.sin(x_data)
y2_data = np.cos(x_data)

# Plotting two lines on the same axes
plt.plot(x_data, y1_data, label="Sine Wave", color="blue", linestyle="-")
plt.plot(x_data, y2_data, label="Cosine Wave", color="red", linestyle="--")

# Adding decorations
plt.title("Sine and Cosine Functions")
plt.xlabel("Angle [rad]")
plt.ylabel("Value")
plt.grid(True)
plt.legend()

plt.show()
```

### The "Clean" Object-Oriented Way

The way we've been plotting is fast and easy. However, for more complex plots (like multiple plots side-by-side), a more structured method is better. This is the "object-oriented" approach.

You start by creating a figure and a set of subplots (axes) with `fig, ax = plt.subplots()`. 
- `fig` is the entire window or page.
- `ax` is the actual plot inside the window.

Instead of calling `plt.plot()`, you call `ax.plot()`. Most function names are similar, but you use `ax.set_title()`, `ax.set_xlabel()`, etc. Notice the `set_` prefix. This method is preferred by many developers because it is more explicit.

Here, the creation of one plot is described. However, subplots easily allows you to create multiple plots next to each other. A tutorial on how this works can be found [here](https://matplotlib.org/stable/gallery/subplots_axes_and_figures/subplots_demo.html).
