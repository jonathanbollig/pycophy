The **print statement** to generate output text:

\# indicate a single line comment

**Variables** are a way to store data:


```python
a = "Hello World"
print(a)
```

`"Hello World"` is a string. We can also print and save other **types of data**:
| Name      | What                     | Example       |
|-----------|--------------------------|---------------|
| String    | Text                     | `"Hello World"` |
| int       | Integer (Number)         | `1`, `10`, `-123`   |
| float     | Rational Number          | `3.14`, `1.99`, `**1.2e-6**`  |
| boolean   | True / False             | `True`, `False`   |
| Emptyness | Explictly saving nothing | `None`          |

Separating arguments in the `print` statement, will show them next to each other.

```python
this_is_a_string = "Hello Mars"
this_is_a_int = 10
this_is_a_float = 9.31
this_is_a_boolean = True
this_is_no_value = None
print(this_is_a_string)
print(this_is_a_int)
print(this_is_a_float)
print(this_is_a_boolean)
print(this_is_no_value)
print()
print(this_is_a_string, this_is_a_int,  this_is_a_float, this_is_a_boolean, this_is_no_value)
```

In python, variables can be **reassigned** as much as one likes. Also reassignments between different types are possible.

```python
b = 1
print(b)
b = "I am now a beautiful string"
print(b)
```

Variables con be modified and combined by **arithmetic operators**:

| Data Type | +              | -         | *        | /          | // (Floor Divide)   | % (Modulo)          | ** (Exponentiate)       |
|-----------|----------------------|---------------------|---------------------|---------------------|---------------------|---------------------|-------------------------|
|`float`     | Valid | Valid               | Valid               | Valid               | Valid (floored float)| Valid               | Valid                   |
|`int`       | Valid   | Valid               | Valid               | Valid (float result) | Valid (int result)  | Valid               | Valid                   |
|`string`    | Concatenation         | Invalid             | Repeat (with int)   | Invalid             | Invalid             | Invalid             | Invalid                 |
|`boolean`   | Treated as int (0,1) | Treated as int      | Treated as int      | Treated as int      | Treated as int      | Treated as int      | Treated as int          |

For numbers (ints and floats) the operators act as expected. However, also strings and booleans can be altered:
- Booleans: Act as `1` (True) and `0` (False)
- Strings: are concatenated with `+`: `"Hello" + "World" = "HelloWorld"` or repeated with `*`: `3*"Hi "="Hi Hi Hi "`

```python
print(3+4, 3-4, 4/4, 3*4, 3**4, 3//4, 3%4)
print("Hello" + "World")
print(3*"Hello " + 3*"World ")
print(False*True, True)
```

The **`type` operator** returns the type of a variabe:

```python
print(type(1))
print(type("Hi"))
print(type(True))
print(type(True*True))
```

**Error messages** appear, if python is not able to execute your code:

```python
impossible_task = "My age is " + 99
```

In this case, the operator "+" does not work with a string and an integer. This is explained in the last error line.

What do we need to fix this: **Type parsing**: Changing a data type. This usually happens automatically:

```python
integer = 1 
print(type(integer))
integer = integer / 2
print(type(integer))
```

But we can also force it:

```python
integer = 1 
print(type(integer))
integer = float(integer)
print(type(integer))
```

This way, we can convert floats or integers to strings: `str()`, and fix the problem at the top:

```python
possible_task = "My age is " + str(99)
print(possible_task)
```
