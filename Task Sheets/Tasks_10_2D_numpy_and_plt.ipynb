**2D NumPy arrays** (matrices) are created from a list of lists. Each sub-list must have the same length.

```python
import numpy as np
import matplotlib.pyplot as plt

my_matrix = np.array([[1, 2, 3], 
                        [4, 5, 6]])
print(my_matrix)
print(f"Shape of the array: {my_matrix.shape}")
```

You can also create 2D arrays using functions like `np.zeros()`, `np.ones()`, or `np.full()` by passing a tuple `(rows, columns)` as the shape.

```python
zeros_matrix = np.zeros((2, 4)) # 2 rows, 4 columns
print("Zeros matrix:")
print(zeros_matrix)

pi_matrix = np.full((3, 2), np.pi) # 3 rows, 2 columns, filled with pi
print("Matrix full of Pi:")
print(pi_matrix)
```

**Indexing and Slicing** in 2D uses a comma to separate the dimensions: `array[row, column]`.

```python
# Get a single element
print(f"Element at row 0, col 2: {my_matrix[0, 2]}")

# Get an entire row (row 0, all columns)
print(f"First row: {my_matrix[0, :]}")

# Get an entire column (all rows, column 1)
print(f"Second column: {my_matrix[:, 1]}")

# Slice a sub-matrix (all rows, columns from 1 onwards)
print(f"Sub-matrix:\n {my_matrix[:, 1:]}")
```

**Visualizing 2D Arrays** is done with `plt.imshow()`. It maps the value of each element in the array to a color.

```python
random_matrix = np.random.random((10, 10)) 

plt.imshow(random_matrix, extent=(0, 1, 0, 1))
plt.title("Heatmap of a 2D Array")
plt.colorbar(label="Value")
plt.show()
```

### Visualizing Fields with Contour Plots

In physics, we often deal with fields (like electric potential, temperature, or pressure) that vary over a 2D space. While `imshow` gives a colored heatmap, a **contour plot** is often used to show lines of constant value (like isobars on a weather map).

To do this, we first need to create a coordinate grid. `np.meshgrid()` is a key function for this. It takes 1D arrays for x and y coordinates and returns two 2D grid arrays, `X` and `Y`. We can then calculate the field value `Z` for each point on the grid. Finally, `plt.contourf()` creates a filled contour plot.
