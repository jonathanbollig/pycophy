**if / else / elif - statements**

```python
condition = True
if condition:
    print("This part is executed, if condition is True")
elif condition == 1:
    print("This part is executed if condition equals one")
else:
    print("This part is printed, if none of the above statements are true")
```

**Condition Operators**
| Operator | Name                     | Example    |
|----------|--------------------------|------------|
| ==       | Equal                    | x == y     |
| !=       | Not equal                | x != y     |
| >        | Greater than             | x > y      |
| <        | Less than                | x < y      |
| >=       | Greater than or equal to | x >= y     |
| <=       | Less than or equal to    | x <= y     |


**Combining Conditions**

(In order of precedence / execution, first to last)
- `not a` --> True if `a` is False
- `a and b` --> True if `a` and `b` are True
- `a or b` --> True if either `a`, `b`, or both are True

### String Manipulation
We have worked a lot with strings so far. However, there are a lot of built-in string functions, that make manipulation easy.

Here is a summary of some string functions. However, there are much more. You can find them [here](https://www.w3schools.com/python/python_ref_string.asp).
| Function        | Description                            | Example                          |
| --------------- | ------------------------------------ | ------------------------------- |
| lower()         | Converts string to lowercase          | `'Hello'.lower() → 'hello'`       |
| upper()         | Converts string to uppercase          | `'hello'.upper() → 'HELLO'`       |
| strip()         | Removes leading/trailing whitespace   | `'  hello  '.strip() → 'hello'`   |
| replace(old, new)| Replaces occurrences of old with new | `'hello world'.replace('world', 'there') → 'hello there'` |
| capitalize()    | Capitalizes first character           | `'hello'.capitalize() → 'Hello'`  |
| title()         | Converts string to title case          |` 'hello world'.title() → 'Hello World'` |


```python
# some examples:
string = "  Hello  "
print(string.upper())
print(string.lower())
print(string.strip())
print(string.replace("l", "y"))
```
