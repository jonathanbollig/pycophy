**Lists**
Lists are used to store multiple items in a single variable. They are created with square brackets `[]`.

````python
# A list can contain different data types, including other lists
measurements = [3.13, 5.83, 7.22]
mixed_list = [3.5, 3j, "hi", True, None]
nested_list = [[1, 3, 5], "hi", True]

print(measurements)
print(mixed_list)
print(nested_list)
````

**Indexing and Slicing**
Access elements of a list using their index. Python is 0-indexed (the first element is at index 0).
- `my_list[0]` → Access the first element
- `my_list[-1]` → Access the last element
- `my_list[1:3]` → Slice from index 1 up to (but not including) index 3
- `my_list[:3]` → Slice from the beginning up to index 3
- `my_list[3:]` → Slice from index 3 to the end

````python
my_list = [100, 50, 0, 99, 4, 42]
print("First element:", my_list[0])
print("Last element:", my_list[-1])
print("Slice from 1 to 3:", my_list[1:3])
````

**Modifying & Operating on Lists**
You can change elements, add elements, and combine lists.

````python
my_list = [100, 50, 0, 99, 4, 42]
print("Original:", my_list)

my_list[0] = 1000000
print("After changing one element:", my_list)

my_list.append(10)
print("After appending 10:", my_list)

list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
print("Concatenated:", list1 + list2)
print("Repeated:", list2 * 3)
````

**Common List Functions**
| Function/Method | Description                            |
| --------------- | ------------------------------------ |
| `len(list)`     | Returns the number of elements.        |
| `list.append(x)`| Adds an element `x` to the end of the list. |
| `max(list)`     | Returns the largest item in a numeric list. |
| `min(list)`     | Returns the smallest item in a numeric list. |
| `sum(list)`     | Returns the sum of a numeric list.     |

````python
li = [1, 43, 2, 3]
print("Length:", len(li))
print("Max value:", max(li))
print("Min value:", min(li))
print("Sum of values:", sum(li))
````

**Tuples**
Tuples are similar to lists but are created with parentheses `()` and are **immutable**, meaning they cannot be changed after creation. You can access their elements, but you cannot reassign them.

```python
matrix = [[1, 2, 3],

```
is a list with three entries. Each entry itself is a list with three integer entries. 

To enter the first element of the matrix, we can:

````python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
first_line = matrix[0]
print(first_line)
````

We can see, that if we access the first element of the matrix, we get a list. As usual, we can again access the first element:

````python
first_element = first_line[0]
print(first_element)
````

However, we can access the `1` element in the matrix directly, by chaining the index: `list[index1][index2]`:

````python
print(matrix[0][0])
````
