**Variables** are a way to store data:


```python
a = "Hello World"
print(a)
```

`"Hello World"` is a string. We can also print and save other **types of data**:
| Name      | What                     | Example       |
|-----------|--------------------------|---------------|
| String    | Text                     | `"Hello World"` |
| int       | Integer (Number)         | `1`, `10`, `-123`   |
| float     | Rational Number          | `3.14`, `1.99`, `1.2e-6`  |
| boolean   | True / False             | `True`, `False`   |
| Emptiness | Explicitly saving nothing | `None`          |

Separating arguments in the `print` statement, will show them next to each other.

```python
this_is_a_string = "Hello Mars"
this_is_a_int = 10
this_is_a_float = 9.31
this_is_a_boolean = True
this_is_no_value = None
print(this_is_a_string, this_is_a_int, this_is_a_float, this_is_a_boolean, this_is_no_value)
```

Variables con be modified and combined by **arithmetic operators**:

| Data Type | +              | -         | *        | /          | // (Floor Divide)   | % (Modulo)          | ** (Exponentiate)       |
|-----------|----------------------|---------------------|---------------------|---------------------|---------------------|---------------------|-------------------------|
|`float`     | Valid | Valid               | Valid               | Valid               | Valid (floored float)| Valid               | Valid                   |
|`int`       | Valid   | Valid               | Valid               | Valid (float result) | Valid (int result)  | Valid               | Valid                   |
|`string`    | Concatenation         | Invalid             | Repeat (with int)   | Invalid             | Invalid             | Invalid             | Invalid                 |
|`boolean`   | Treated as int (0,1) | Treated as int      | Treated as int      | Treated as int      | Treated as int      | Treated as int      | Treated as int          |

For numbers (ints and floats) the operators act as expected. However, also strings and booleans can be altered:
- Booleans: Act as `1` (True) and `0` (False)
- Strings: are concatenated with `+`: `"Hello" + "World" = "HelloWorld"` or repeated with `*`: `3*"Hi "="Hi Hi Hi "`

The **`type` operator** returns the type of variable: `type("Hi") --> str`

**Type parsing**: Changing a data type. This often happens automatically, but we can also force it:

```python
integer = 1 
print(type(integer))
integer = float(integer)
print(type(integer))
```

### Rounding Numbers
The `round()` operator rounds integers. A second parameter can be passed, to determine the decimal precision:

```python
number = 5/3
print(number)
print(round(number))
print(round(number, 2))
print(round(number, 4))
```

### Advanced Data Types
Besides the introduced data types, python has a few more. Lists, Sets, Tuples and Dictionaries are introduced later. However, here I want to introduce **complex numbers**, which have their own data type in python. Complex numbers are denoted with a "j" after the number: *3+4j* or *3.14+0.1j*.

```python
print(3+4j)

# square root of -1:
print((-1)**(1/2))
```

### Naming Conventions

[PEP 8](https://peps.python.org/pep-0008/) is a very extensive styling guideline for python code. 
The style of your code does not affect execution of your code. However, it is important to write code, that is nicely readable by other people. Therefore, it is a good idea to stuck to conventions:
- Introduced in lecture: Alphanumerical + "_"
- lowercase, separated by underscores
- Variable names should be descriptive, concise, and meaningful

Good: `user_name = ...`

Bad: `usname = ...`, `UserName = ...`, `USER_NAME = ...`
